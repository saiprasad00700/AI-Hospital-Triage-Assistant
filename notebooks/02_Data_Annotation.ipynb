```python
import pandas as pd
from datasets import load_dataset

# Load MedDialog
dataset = load_dataset("OpenMed/MedDialog")

# Convert train split to pandas
train_df = dataset["train"].to_pandas()

# Remove duplicates
train_df = train_df.drop_duplicates().reset_index(drop=True)

train_df.shape
```

```python
INTENTS = [
    "Symptom",
    "Appointment",
    "Prescription",
    "Lab_Report",
    "Emergency",
    "Hospital_Info"
]

INTENTS
```

```python
EMERGENCY_KEYWORDS = [
    "heart attack",
    "chest pain",
    "cannot breathe",
    "can't breathe",
    "difficulty breathing",
    "shortness of breath",
    "unconscious",
    "severe bleeding",
    "stroke",
    "collapsed"
]

APPOINTMENT_KEYWORDS = [
    "appointment",
    "book appointment",
    "schedule",
    "consultation",
    "visit doctor",
    "meet doctor"
]

PRESCRIPTION_KEYWORDS = [
    "prescription",
    "refill",
    "medicine",
    "medication",
    "tablet",
    "drug"
]

LAB_REPORT_KEYWORDS = [
    "blood report",
    "lab report",
    "test report",
    "x-ray",
    "mri",
    "ct scan"
]

HOSPITAL_INFO_KEYWORDS = [
    "hospital address",
    "location",
    "timing",
    "opening hours",
    "contact number"
]
```

```python
def annotate_intent(text):

    text = str(text).lower()

    if any(word in text for word in EMERGENCY_KEYWORDS):
        return "Emergency"

    elif any(word in text for word in APPOINTMENT_KEYWORDS):
        return "Appointment"

    elif any(word in text for word in PRESCRIPTION_KEYWORDS):
        return "Prescription"

    elif any(word in text for word in LAB_REPORT_KEYWORDS):
        return "Lab_Report"

    elif any(word in text for word in HOSPITAL_INFO_KEYWORDS):
        return "Hospital_Info"

    else:
        return "Symptom"
```

```python
# Apply improved annotation
train_df["intent"] = train_df["patient_message"].apply(annotate_intent)

# Preview the results
train_df[["patient_message", "intent"]].head(10)
```

```python
intent_counts = train_df["intent"].value_counts()

print(intent_counts)
```

```python
# Save annotated dataset
train_df.to_csv("annotated_meddialog.csv", index=False)

print("Annotated dataset saved successfully!")
```
