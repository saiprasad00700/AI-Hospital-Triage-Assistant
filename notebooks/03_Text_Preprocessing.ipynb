```python
import pandas as pd
import re
from datasets import load_dataset

# Load dataset
dataset = load_dataset("OpenMed/MedDialog")

train_df = dataset["train"].to_pandas()

# Remove duplicates
train_df = train_df.drop_duplicates().reset_index(drop=True)

train_df.shape
```

```python
train_df = train_df[["patient_message"]]

train_df.head()
```

```python
train_df["clean_text"] = train_df["patient_message"]
```

```python
train_df["clean_text"] = train_df["clean_text"].str.lower()
```

```python
# Remove newline, carriage return and tab characters
train_df["clean_text"] = train_df["clean_text"].str.replace(r"[\n\r\t]", " ", regex=True)

train_df[["patient_message", "clean_text"]].head(3)
```

```python
# Remove email addresses
train_df["clean_text"] = train_df["clean_text"].str.replace(
    r"\S+@\S+", " ", regex=True
)

# Remove phone numbers (10 or more digits)
train_df["clean_text"] = train_df["clean_text"].str.replace(
    r"\b\d{10,}\b", " ", regex=True
)

# Remove URLs
train_df["clean_text"] = train_df["clean_text"].str.replace(
    r"http\S+|www\S+", " ", regex=True
)

train_df[["patient_message", "clean_text"]].head(5)
```

```python
# Keep only letters, numbers and spaces
train_df["clean_text"] = train_df["clean_text"].str.replace(
    r"[^a-z0-9\s]",
    " ",
    regex=True
)

train_df[["patient_message", "clean_text"]].head(5)
```

```python
# Replace multiple spaces with a single space
train_df["clean_text"] = train_df["clean_text"].str.replace(
    r"\s+",
    " ",
    regex=True
).str.strip()

train_df[["patient_message", "clean_text"]].head(5)
```

```python
import re

# Create editable copy
train_df["clean_text"] = train_df["patient_message"]

# Lowercase
train_df["clean_text"] = train_df["clean_text"].str.lower()

# Remove newlines
train_df["clean_text"] = train_df["clean_text"].str.replace(r"[\n\r\t]", " ", regex=True)

# Remove emails
train_df["clean_text"] = train_df["clean_text"].str.replace(r"\S+@\S+", " ", regex=True)

# Remove phone numbers
train_df["clean_text"] = train_df["clean_text"].str.replace(r"\b\d{10,}\b", " ", regex=True)

# Remove URLs
train_df["clean_text"] = train_df["clean_text"].str.replace(r"http\S+|www\S+", " ", regex=True)

# Remove punctuation
train_df["clean_text"] = train_df["clean_text"].str.replace(r"[^a-z0-9\s]", " ", regex=True)

# Remove extra spaces
train_df["clean_text"] = train_df["clean_text"].str.replace(r"\s+", " ", regex=True).str.strip()
```

```python
processed_df = train_df[["clean_text", "intent"]]

processed_df.head()
```
